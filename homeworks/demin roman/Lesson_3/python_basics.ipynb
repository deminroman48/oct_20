```python
"""
================================================================
ПОЛНЫЙ КОНСПЕКТ ПО ОСНОВАМ PYTHON ДЛЯ НОВИЧКА
================================================================
Этот файл можно запустить (python file.py) — все примеры рабочие.
Формат: комментарии (#) — теория, код — примеры.
================================================================
"""

# ==============================================================
# 1. ПЕРЕМЕННЫЕ И ТИПЫ ДАННЫХ
# ==============================================================
# Python — динамически типизированный язык: тип определяется сам.

name = "Алиса"       # str (строка)
age = 25             # int (целое число)
height = 1.68        # float (дробное число)
is_student = True    # bool (логический тип)
nothing = None       # NoneType (отсутствие значения)

# Узнать тип:
print(type(age))     # <class 'int'>

# Приведение типов:
print(int("42"))     # 42
print(str(100))      # "100"
print(float("3.14")) # 3.14
print(bool(0))       # False


# ==============================================================
# 2. ОПЕРАТОРЫ
# ==============================================================
# Арифметические:
print(7 + 2)    # 9
print(7 - 2)    # 5
print(7 * 2)    # 14
print(7 / 2)    # 3.5   — всегда float
print(7 // 2)   # 3     — целочисленное деление
print(7 % 2)    # 1     — остаток
print(7 ** 2)   # 49    — степень

# Сравнения: ==  !=  >  <  >=  <=
# Логические: and, or, not
# Принадлежность: in, not in


# ==============================================================
# 3. СТРОКИ (str)
# ==============================================================
s = "Привет, мир!"

print(len(s))            # длина
print(s.upper())         # ПРИВЕТ, МИР!
print(s.lower())         # привет, мир!
print(s.strip())         # убирает пробелы по краям
print(s.replace("мир", "Python"))
print(s.split(", "))     # ['Привет', 'мир!']
print(s[0])              # 'П' (индексация с 0)
print(s[-1])             # '!' (с конца)
print(s[0:6])            # 'Привет' (срез)

# f-строки (современный способ форматирования):
name = "Иван"
print(f"Привет, {name}! Тебе {age} лет.")

# Многострочная строка:
text = """Первая строка
Вторая строка"""


# ==============================================================
# 4. СПИСКИ (list) — изменяемые упорядоченные коллекции
# ==============================================================
fruits = ["яблоко", "банан", "вишня"]

fruits.append("груша")       # добавить в конец
fruits.insert(1, "киви")     # вставить по индексу
fruits.remove("банан")       # удалить по значению
last = fruits.pop()          # удалить и вернуть последний
print(fruits[0])             # 'яблоко'
print(fruits[-1])            # последний
print(fruits[1:3])           # срез
print(len(fruits))           # количество
print("киви" in fruits)      # True

# Перебор:
for f in fruits:
    print(f)

# Сортировка:
numbers = [3, 1, 4, 1, 5]
numbers.sort()               # [1, 1, 3, 4, 5]
numbers.sort(reverse=True)   # по убыванию


# ==============================================================
# 5. КОРТЕЖИ (tuple) — неизменяемые
# ==============================================================
point = (10, 20)
x, y = point                 # распаковка
print(point[0])              # 10


# ==============================================================
# 6. СЛОВАРИ (dict) — пары ключ: значение
# ==============================================================
person = {"name": "Мария", "age": 30}

person["city"] = "Москва"    # добавление
print(person["name"])        # 'Мария'
print(person.get("job", "—"))# значение или значение по умолчанию
del person["age"]            # удаление

for key, value in person.items():
    print(key, "=", value)


# ==============================================================
# 7. МНОЖЕСТВА (set) — уникальные значения, без порядка
# ==============================================================
nums = {1, 2, 3, 2, 1}
print(nums)                  # {1, 2, 3}

a = {1, 2, 3}
b = {3, 4, 5}
print(a | b)                 # объединение {1, 2, 3, 4, 5}
print(a & b)                 # пересечение {3}
print(a - b)                 # разность {1, 2}


# ==============================================================
# 8. УСЛОВИЯ (if / elif / else)
# ==============================================================
score = 85

if score >= 90:
    grade = "A"
elif score >= 75:
    grade = "B"
else:
    grade = "C"

print(grade)  # B

# Тернарный оператор:
status = "взрослый" if age >= 18 else "ребёнок"


# ==============================================================
# 9. ЦИКЛЫ
# ==============================================================
# for — перебор последовательности
for i in range(5):           # 0, 1, 2, 3, 4
    print(i)

for i in range(2, 10, 2):    # от 2 до 10 с шагом 2
    print(i)

# while — пока условие истинно
count = 0
while count < 3:
    print(count)
    count += 1

# break — прервать, continue — перейти к следующей итерации
for i in range(10):
    if i == 5:
        break
    if i % 2 == 0:
        continue
    print(i)


# ==============================================================
# 10. ФУНКЦИИ
# ==============================================================
def greet(name, greeting="Привет"):
    """Возвращает приветствие."""
    return f"{greeting}, {name}!"

print(greet("Олег"))              # Привет, Олег!
print(greet("Олег", "Здравствуй"))

# Функция с переменным числом аргументов:
def total(*args, **kwargs):
    print(args)      # кортеж позиционных
    print(kwargs)    # словарь именованных
    return sum(args)

total(1, 2, 3, x=10)


# ==============================================================
# 11. ВВОД / ВЫВОД
# ==============================================================
# Ввод (раскомментируй для теста):
# user_name = input("Как тебя зовут? ")
# print(f"Привет, {user_name}!")

# Вывод с разделителями:
print("a", "b", "c", sep="-")    # a-b-c
print("без переноса", end=" ")
print("на новой строке")


# ==============================================================
# 12. ОБРАБОТКА ИСКЛЮЧЕНИЙ (try / except)
# ==============================================================
try:
    result = 10 / 0
except ZeroDivisionError as e:
    print("Ошибка:", e)
finally:
    print("Этот блок выполнится всегда")


# ==============================================================
# 13. ГЕНЕРАТОРЫ СПИСКОВ (list comprehension)
# ==============================================================
squares = [x**2 for x in range(10)]
evens = [x for x in range(20) if x % 2 == 0]
words = {word: len(word) for word in ["a", "bb", "ccc"]}


# ==============================================================
# 14. МОДУЛИ (import)
# ==============================================================
import math
from random import randint
import os

print(math.sqrt(16))             # 4.0
print(math.pi)                   # 3.1415...
print(randint(1, 10))            # случайное число
print(os.getcwd())               # текущая папка


# ==============================================================
# 15. ПОЛЕЗНЫЕ ВСТРОЕННЫЕ ФУНКЦИИ
# ==============================================================
print(len([1, 2, 3]))            # длина
print(max(1, 5, 3))              # максимум
print(min(1, 5, 3))              # минимум
print(sum([1, 2, 3]))            # сумма
print(abs(-7))                   # модуль
print(round(3.14159, 2))         # округление
print(sorted([3, 1, 2]))         # новая сортированная копия
print(list(reversed([1, 2, 3])))
print(enumerate(["a", "b"]))     # пары (индекс, значение)
print(zip([1, 2], ["a", "b"]))   # объединение


# ==============================================================
# 16. КЛАССЫ (кратко)
# ==============================================================
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def bark(self):
        return f"{self.name} говорит: Гав!"

rex = Dog("Рекс", 3)
print(rex.bark())


# ==============================================================
# 17. СТИЛЬ КОДА (PEP 8) — кратко
# ==============================================================
# ✔ имена переменных и функций — snake_case: my_variable
# ✔ имена классов — CamelCase: MyClass
# ✔ константы — UPPER_CASE: MAX_SIZE
# ✔ отступы — 4 пробела (не табы!)
# ✔ строки — до ~79 символов
# ✔ после запятой в списках — пробел: [1, 2, 3]
# ✔ операторы окружаются пробелами: a + b
# ✔ пустые строки: 2 между функциями/классами
# ✔ комментарии: # пробел после решётки


# ==============================================================
# ИТОГ: с чем дальше знакомиться
# ==============================================================
# • Работа с файлами (open, with)
# • Декораторы и генераторы (yield)
# • ООП углублённо (наследование, магические методы)
# • Стандартная библиотека (pathlib, json, datetime, re)
# • Популярные библиотеки: requests, numpy, pandas
# • Виртуальные окружения: python -m venv .venv
# • Типизация: mypy, аннотации (def f(x: int) -> str: ...)

print("\n✅ Все примеры выше — рабочий код. Измени их и экспериментируй!")
```
