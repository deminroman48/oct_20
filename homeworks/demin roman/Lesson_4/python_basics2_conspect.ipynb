## 1. Операторы составного присвоения

Составное присвоение объединяет арифметическую операцию и присваивание: `x op= y` эквивалентно `x = x op y`.

| Оператор | Пример | Эквивалент |
|---|---|---|
| `+=` | `x += 3` | `x = x + 3` |
| `-=` | `x -= 2` | `x = x - 2` |
| `*=` | `x *= 4` | `x = x * 4` |
| `/=` | `x /= 2` | `x = x / 2` |
| `//=` | `x //= 3` | `x = x // 3` |
| `%=` | `x %= 3` | `x = x % 3` |
| `**=` | `x **= 2` | `x = x ** 2` |
| `&=` | `x &= y` | `x = x & y` |
| `\|=` | `x \|= y` | `x = x \| y` |
| `^=` | `x ^= y` | `x = x ^ y` |
| `>>=` | `x >>= 1` | `x = x >> 1` |
| `<<=` | `x <<= 1` | `x = x << 1` |

```python
x = 10
x += 5          # 15
print("x += 5   ->", x)

x -= 3          # 12
print("x -= 3   ->", x)

x *= 2          # 24
print("x *= 2   ->", x)

x //= 5         # 4
print("x //= 5  ->", x)

x **= 3         # 64
print("x **= 3  ->", x)

x %= 10         # 4
print("x %= 10  ->", x)
```

## 2. Операторы принадлежности

Операторы `in` и `not in` проверяют, содержится ли элемент в последовательности или коллекции (строка, список, кортеж, множество, словарь по ключам). Результат — `bool`.

- `x in s` — `True`, если `x` есть в `s`.
- `x not in s` — `True`, если `x` отсутствует в `s`.

```python
print("'a' in 'abc'", 'a' in 'abc')          # True
print("'z' not in 'abc'", 'z' not in 'abc')  # True

numbers = [10, 20, 30]
print("20 in numbers", 20 in numbers)        # True
print("40 not in numbers", 40 not in numbers) # True

person = {'name': 'Alice', 'age': 30}
print("'name' in person", 'name' in person)   # True (проверка ключа)
print("'Alice' in person", 'Alice' in person) # False (значение не проверяется)
```

## 3. Операторы идентичности

`is` и `is not` проверяют, являются ли две переменные ссылками на **один и тот же объект в памяти** (сравнение `id()`), а не равны ли их значения.

- `x is y` → `True`, если `id(x) == id(y)`.
- `x is not y` → `True`, если `id(x) != id(y)`.

> **Важно:** для сравнения значений используйте `==`. `is` — только для проверки `None`, `True`, `False` и подобных синглтонов.

```python
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print("a == b:", a == b)    # True — значения равны
print("a is b:", a is b)    # False — разные объекты
print("a is c:", a is c)    # True — c ссылается на тот же объект

x = None
print("x is None:", x is None)  # правильный способ проверки на None
```

## 4. Способы ввода данных

Основные способы получить данные от пользователя или из стандартного потока:

1. **`input(prompt)`** — читает одну строку, всегда возвращает `str`.
2. **`sys.stdin.readline()`** — чтение строки из стандартного ввода (аналогично `input`, но без автоматического удаления `\n`).
3. **`sys.stdin.read()`** — читает весь ввод до конца файла (удобно для пакетной обработки).
4. **`for line in sys.stdin:`** — построчное чтение (эффективно для больших объёмов).
5. **`fileinput.input()`** — универсальное чтение из файлов, переданных в аргументах, или из `stdin`.

```python
# 1. input() — самый простой способ
name = input("Введите имя: ")
print(f"Привет, {name}!")

# Если нужны числа — преобразуем явно
age = int(input("Введите возраст: "))
print(f"Через год вам будет {age + 1}")
```

## 5. Преобразование типов данных

Python поддерживает **неявное** (автоматическое) и **явное** (ручное) преобразование.

- **Неявное:** `int + float → float`, `bool + int → int`.
- **Явное:** функции `int()`, `float()`, `str()`, `bool()`, `list()`, `tuple()`, `set()`, `dict()`.

> При явном преобразовании возможны потери данных или ошибки (`ValueError`, `TypeError`).

```python
# Числовые преобразования
print(int("42"))          # 42
print(float("3.14"))      # 3.14
print(int(3.99))          # 3 — дробная часть отбрасывается
print(str(123))           # '123'
print(bool(0), bool(1))   # False True

# Коллекции
print(list("abc"))        # ['a', 'b', 'c']
print(tuple([1, 2, 3]))   # (1, 2, 3)
print(set([1, 2, 2, 3]))  # {1, 2, 3}
print(dict([("a", 1), ("b", 2)]))  # {'a': 1, 'b': 2}
```

## 6. 8 типов данных в Python

В рамках базового курса обычно выделяют следующие 8 ключевых типов:

| № | Тип | Пример | Описание |
|---|---|---|---|
| 1 | `int` | `42` | Целые числа произвольной длины |
| 2 | `float` | `3.14` | Вещественные числа (IEEE 754) |
| 3 | `complex` | `3+4j` | Комплексные числа |
| 4 | `str` | `"hello"` | Unicode-строки |
| 5 | `list` | `[1, 2, 3]` | Изменяемая упорядоченная последовательность |
| 6 | `tuple` | `(1, 2, 3)` | Неизменяемая упорядоченная последовательность |
| 7 | `set` | `{1, 2, 3}` | Неупорядоченное множество уникальных элементов |
| 8 | `dict` | `{"a": 1, "b": 2}` | Словарь «ключ → значение» |

> Дополнительно существуют `bool` (наследник `int`), `NoneType` (`None`), `bytes`, `frozenset`, `range` и др.

```python
examples = [
    (42, int),
    (3.14, float),
    (3 + 4j, complex),
    ("hello", str),
    ([1, 2, 3], list),
    ((1, 2, 3), tuple),
    ({1, 2, 3}, set),
    ({"a": 1, "b": 2}, dict),
]

for value, expected_type in examples:
    print(f"{value!r:<20} -> {type(value).__name__:<8} (ожидался {expected_type.__name__})")
```

## 7. Списки (`list`)

**Список** — изменяемая упорядоченная коллекция, допускающая дубликаты и элементы разных типов.

**Создание:** `[]`, `list()`, списковое включение.

**Основные методы:**

| Метод | Действие |
|---|---|
| `append(x)` | Добавить `x` в конец |
| `extend(iterable)` | Добавить все элементы итерируемого объекта |
| `insert(i, x)` | Вставить `x` на позицию `i` |
| `remove(x)` | Удалить первое вхождение `x` |
| `pop([i])` | Удалить и вернуть элемент по индексу (по умолчанию последний) |
| `index(x)` | Индекс первого вхождения `x` |
| `count(x)` | Количество вхождений `x` |
| `sort()` / `reverse()` | Сортировка на месте / разворот |
| `copy()` | Поверхностная копия |

```python
fruits = ["apple", "banana", "cherry"]
fruits.append("orange")
fruits.insert(1, "kiwi")
print(fruits)

fruits.remove("banana")
last = fruits.pop()
print(fruits, "| удалён:", last)

numbers = [3, 1, 4, 1, 5]
numbers.sort()
print(numbers)
print("count(1):", numbers.count(1))
```

## 8. Кортежи (`tuple`)

**Кортеж** — неизменяемая упорядоченная коллекция. После создания элементы нельзя добавить, удалить или заменить.

**Создание:** `()`, `tuple()`, просто перечисление через запятую.

**Особенности:**
- Методов изменения нет.
- Доступны `count()` и `index()`.
- Может использоваться как ключ словаря (если все элементы хешируемы).
- Распаковка: `a, b = (1, 2)`.

```python
point = (10, 20)
print(point[0], point[1])       # 10 20
print("count(10):", point.count(10))
print("index(20):", point.index(20))

x, y = point                    # распаковка
print("x =", x, "y =", y)

# point[0] = 99  # TypeError: 'tuple' object does not support item assignment
```

## 9. Множества (`set`)

**Множество** — неупорядоченная коллекция уникальных элементов. Дубликаты автоматически удаляются.

**Создание:** `{1, 2, 3}`, `set()`. Пустое множество — только `set()`, потому что `{}` создаёт словарь.

**Основные методы:**

| Метод | Действие |
|---|---|
| `add(x)` | Добавить элемент |
| `remove(x)` / `discard(x)` | Удалить элемент (`remove` бросает `KeyError`, если элемента нет) |
| `union(other)` / `\|` | Объединение |
| `intersection(other)` / `&` | Пересечение |
| `difference(other)` / `-` | Разность |
| `symmetric_difference(other)` / `^` | Симметричная разность |
| `update(other)` | Добавить все элементы из `other` |

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print("Объединение:", a | b)               # {1, 2, 3, 4, 5, 6}
print("Пересечение:", a & b)              # {3, 4}
print("Разность a - b:", a - b)           # {1, 2}
print("Симметричная разность:", a ^ b)    # {1, 2, 5, 6}

a.add(10)
a.discard(1)
print("После изменений:", a)
```

## 10. Словари (`dict`)

**Словарь** — изменяемая коллекция пар «ключ: значение». Начиная с Python 3.7 сохраняет порядок добавления. Ключи должны быть хешируемыми и уникальными.

**Создание:** `{}`, `dict()`, `{key: value for ...}`.

**Основные операции и методы:**

| Операция / метод | Действие |
|---|---|
| `d[key] = value` | Добавить или обновить пару |
| `d[key]` | Получить значение (иначе `KeyError`) |
| `d.get(key, default)` | Безопасно получить значение |
| `del d[key]` | Удалить пару |
| `key in d` | Проверить наличие ключа |
| `keys()` / `values()` / `items()` | Представления ключей / значений / пар |
| `pop(key)` | Удалить и вернуть значение |
| `update(other)` | Обновить словарь парами из `other` |

```python
student = {"name": "Alice", "age": 22}
student["grade"] = "A"          # добавление
student["age"] = 23             # обновление
print(student)

print("Имя:", student["name"])
print("Курс (нет ключа):", student.get("course", "не указан"))

print("Ключи:", list(student.keys()))
print("Значения:", list(student.values()))
print("Пары:", list(student.items()))

del student["grade"]
print("После удаления:", student)
```
